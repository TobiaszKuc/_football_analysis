```python
import pandas as pd
import matplotlib.pyplot as plt
pd.options.display.max_columns = None
```

```python
mecz = pd.read_csv("data/Polonia Bytom_Pogo  Grodzisk Mazowiecki_4068759.csv")
```

```python
mecz.head(2)
```

```python
strzaly = mecz[(mecz["event_type_name"] == "Shot") & (mecz["freeze_frame_player_id"].isna() == True)]
```

Sprawdzenie, ile było goli i czy pokrywają się z faktycznym wynikiem.

```python
gole = mecz[(mecz["event_type_name"] == "Shot") & (mecz["outcome_name"] == "Goal") & (mecz["freeze_frame_player_id"].isna() == True)]
gole
```

Funkcja przypisująca wartości w zależności od tego, kto strzelił gola, z perspektywy Pogoni.

```python
strzaly["gole"] = strzaly.apply(
    lambda x: 1 if x["possession_team_name"] == "Pogoń Grodzisk Mazowiecki" and x["outcome_name"] == "Goal"
    else -1 if x["possession_team_name"] == "Polonia Bytom" and x["outcome_name"] == "Goal"
    else 0,
    axis=1)

# Metoda .cumsum() monitoruje na bieżąco stan meczu, co będzie potrzebne do następnej funkcji
strzaly["stan_meczu"] = strzaly["gole"].cumsum()
strzaly["stan_meczu"].head(20)
```

Przypisanie każdemu strzałowi w jakim okresie padł a następnie pogrupowanie po drużynie i okresie oraz policzenie xG.

```python
strzaly["korzysc"] = strzaly.apply(lambda x: "Pogoń wygrywa" if x["stan_meczu"] > 0 
                                   else "Pogoń przegrywa" if x["stan_meczu"] < 0 
                                   else "Remis",
                                   axis = 1)
xG_na_druzyne = strzaly.groupby(["korzysc", "possession_team_name"]).agg({"statsbomb_xg":"sum", "event_type_name":"count"})
xG_na_druzyne = xG_na_druzyne.reset_index()
xG_na_druzyne["xG_nastrzal"] = xG_na_druzyne["statsbomb_xg"]/xG_na_druzyne["event_type_name"]
xG_na_druzyne
```

Stworzenie tabeli, która pozwoli policzyć różnicę xG między drużynami.

```python
roznica_xg = xG_na_druzyne.pivot(index = "korzysc", columns = "possession_team_name", values = "statsbomb_xg")
roznica_xg.columns = ["xG_Pogon", "xG_Polonia"]
roznica_xg = roznica_xg.reset_index()
roznica_xg["roznica_xg"] = roznica_xg["xG_Pogon"] - roznica_xg["xG_Polonia"]
roznica_xg = roznica_xg.sort_values(by = "roznica_xg", ascending = True)
roznica_xg
```

```python
plt.bar(x = roznica_xg["korzysc"], height = roznica_xg["roznica_xg"])
plt.title(label = "Różnica xG pomiędzy drużynami w zależności od stanu meczu")
plt.ylabel("Różnica xG")
```

Dodatkowo zdecydowałem się sprawdzić, jak rozłoży się xG na strzał, ponieważ dla remisu bardzo mocno przeważa ilość strzałów, która jest 1.7 razy wyższa.

```python
roznica_xg_nastrzal = xG_na_druzyne.pivot(index = "korzysc", columns = "possession_team_name", values = "xG_nastrzal")
roznica_xg_nastrzal.columns = ["xG_Pogon", "xG_Polonia"]
roznica_xg_nastrzal = roznica_xg_nastrzal.reset_index()
roznica_xg_nastrzal["roznica_xg_nastrzal"] = roznica_xg_nastrzal["xG_Pogon"] - roznica_xg_nastrzal["xG_Polonia"]
roznica_xg_nastrzal = roznica_xg_nastrzal.sort_values(by = "roznica_xg_nastrzal", ascending = True)
roznica_xg_nastrzal
```

```python
plt.bar(x = roznica_xg_nastrzal["korzysc"], height = roznica_xg_nastrzal["roznica_xg_nastrzal"])
plt.title(label = "Różnica xG na strzał pomiędzy drużynami w zależności od stanu meczu")
plt.ylabel("Różnica xG")
```
