```python
data["average"] = (data["pace"]*0.4 + data["shooting"]*0.35 + data["dribbling"]*0.25)
```

```python
data[["short_name", "pace", "shooting", "dribbling", "average"]].head(10)
```

```python
data["dribbling_average"] = (data["dribbling"]*0.35 + data["skill_ball_control"]*0.65)
```

```python
data["shooting_average"] = (data["shooting"]*0.25 + data["mentality_composure"]*0.35 + data["attacking_finishing"]*0.4)
```

```python
data["pace_average"] = (data["movement_agility"]*0.25 + data["movement_sprint_speed"]*0.4 + data["movement_acceleration"]*0.35)
```

```python
data[(data["short_name"] == "M. Salah")][["short_name", "dribbling_average", "shooting_average", "pace_average"]]
```

```python
data["overall_average_salah"] = (data["shooting_average"]*0.45 + data["pace_average"]*0.35 + data["dribbling_average"]*0.2)
```

```python
data[(data["short_name"] == "M. Salah")][["short_name", "overall_average_salah"]]
```

```python
salah_shortlist = data[(data["overall_average_salah"] >= 80) 
    & (data["player_positions"].str.contains(r"\bRW\b") 
        | data["player_positions"].str.contains(r"\bLW\b")) 
    & (data["wage_eur"] <= 150000) 
    & (data["age"] <= 25)][["short_name", "player_positions", "age", "wage_eur", "overall_average_salah", "shooting_average", "pace_average", "dribbling_average", "club_position"]].sort_values("overall_average_salah",ascending=False)
```

```python
colors = []
for _, row in salah_shortlist.iterrows():
    if row["club_position"] == "LW":
        colors.append("red")
    else:
        colors.append("green")
plt.scatter(x=salah_shortlist["age"], y=salah_shortlist["overall_average_salah"], s=salah_shortlist["wage_eur"]/100, c=colors)
for _, row in salah_shortlist.iterrows():
    plt.annotate(row["short_name"], (row["age"], row["overall_average_salah"]))
    plt.title("Potential Salah replacement")
    plt.xlabel("Age")
    plt.ylabel("Average of shooting, pace and dribbling")
```
