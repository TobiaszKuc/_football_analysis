```python
import matplotlib.pyplot as plt
```

```python
team_counts = df['team_name'].value_counts()
team_counts.plot(kind='bar')
plt.title('Liczba zdarzeń na drużynę')
plt.ylabel('Liczba zdarzeń')
plt.show
```

```python
total_events = len(df)
only_passes = len(df[df['type'] == 'Pass'])
podania_wszystko = (only_passes/total_events)*100
print(podania_wszystko)
```

```python
total_events = len(df)
rozne_eventy = df.groupby('team_name')['type'].value_counts()
jaki_procent_wszystkich = (rozne_eventy/total_events)*100
jaki_procent_wszystkich.plot(kind='bar')
plt.xlabel('Drużyna, rodzaj akcji')
plt.ylabel('Procent wszystkich akcji')
```
