```python
import pandas as pd
import json
```

```python
df = pd.read_json("data/pierwszy_plik.json")
```

```python
df.head(5)
```

```python
df['team_name'] = df['possession_team'].apply(lambda x: x['name'] if isinstance(x, dict) else None)
df['possession_type'] = df['type'].apply(lambda x: x['name'] if isinstance(x,dict) else None)
df['pojedyncze_akcje'] = df.groupby('team_name')['possession'].nunique()
df['wszystkie_eventy'] = df.groupby('team_name')['possession'].size()
```

```python
#bierze nazwe z druzyny z dict ktory ma tam druzyne i id
df['team_name'] = df['possession_team'].apply(lambda x: x['name'] if isinstance(x, dict) else None)
#grupuje po possession, a następnie pokazuje jak dużo było akcji w jednym posiadaniu
akcja_na_posiadanie = df.groupby(['team_name', 'possession']).size()
#grupuje po druzynie, potem usrednia akcje na jedno posiadanie
akcja_na_posiadanie.groupby('team_name').mean()
```

```python
#bierze nazwe akcji
df['possession_type'] = df['type'].apply(lambda x: x['name'] if isinstance(x,dict) else None)
#chcę uzyskać unikalne wartości dla każdej akcji
jakie_akcje_na_posiadanie = df.groupby(['possession_type', 'possession', 'team_name']).size()
#zlicza srednia na kazda akcje przez posiadania
jakie_akcje_na_posiadanie.groupby(['possession_type', 'team_name']).mean()
```

```python
#tutaj chce sprawdzic jaki jest udział procentowy każdej akcji w całości posiadania danej drużyny
#bierze nazwe akcji
df['possession_type'] = df['type'].apply(lambda x: x['name'] if isinstance(x,dict) else None)
#zliczam ilość danej akcji na drużynę
akcje_na_druzyne = df.groupby(['team_name', 'possession_type']).size()
#zilczam sume akcji na druzyne
suma_na_druzyne = df.groupby('team_name').size()
#dziele jedno przez drugie i mam procentowy udział akcji
wynik = round((akcje_na_druzyne/suma_na_druzyne)*100, 2)
print(wynik)
```
